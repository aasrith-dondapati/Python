```python
import numpy as np
```

```python
a=np.array([1,2,3]) # create an array from a list
print('Array a:',a)
```

```python
# Create an array with evenly spaced values
b=np.arange(0,10,2) # Values from 0-10 with step count 2
print('Array b:',b)
```

```python
# Create an array filled with ones
e=np.ones((3,2)) # 3x2 array of ones
print('Array e:\n',e)
```

```python
# Create an idendtity matrix
f=np.eye(3) # 3x3 identity matrix
print('Identity Matrix f:\n',f)
```

Array Manipulation Functions

```python
# Reshape an array
a1=np.array([1,2,3])
reshaped=np.reshape(a1,(1,3)) # Reshape to 1x3 matrix
print('Reshaped Array:',reshaped)
```

```python
# Flatten an array
f1=np.array([[1,2],[3,4]])
flattend=np.ravel(f1) # Flattend to 1D array
print('Flattend Array:',flattend)
```

```python
# Transpose an array
e1=np.array([[1,2],[3,4]])
transposed=np.transpose(e1) # Transpose array
print('Transposed Array e1:\n',transposed)
```

```python
# Stack Arrays Vertically
a2=np.array([1,2])
b2=np.array([3,4])
stacked=np.vstack([a2,b2]) # stack a and b verticalically
print('Stacked Array:\n', stacked)
```

Statistical Functions

```python
s=np.array([1,2,3,4])
mean=np.mean(s) # Average or Mean of s
print('Mean of s:',mean)
```

```python
# Standard deviation of s
std_dev=np.std(s)
print('Standard Deviation of s:',std_dev)
```

```python
#minimum elemet of an array
min=np.min(s)
print('Min of s:',min)
```

```python
#maximum elemet of an array
max=np.max(s)
print('Max of s:',max)
```

Linear Algebra Functions

```python
# Create a matrix
mat=np.array([[1,2],[3,4]])
mat
```

```python
# Determinant of a matrix
determinant=np.linalg.det(mat)
print('Determinent of Matrix:',determinant)
```

Boolean and Logical Functions

```python
#Check if all elements are True
#all
logical_test=np.array([True,False,True])
all_true=np.all(logical_test)
print('All elements True:',all_true)
```

```python
logical_test=np.array([True,True,True])
all_true=np.all(logical_test)
print('All elements True:',all_true)
```

```python
logical_test=np.array([False,False,False])
all_true=np.all(logical_test)
print('All elements True:',all_true)
```

```python
# Check if any elemets are True
logical_test=np.array([True, False,True])
any_true=np.any(logical_test)
print('All elements True:',any_true)
```

Set Operations

```python
# Intersection of two arrays
set_a=np.array([1,2,3,4])
set_b=np.array([3,4,5,6])
intersection=np.intersect1d(set_a,set_b)
print('Intersection of a and b:',intersection)
```

```python
#Union of Two arrays
union=np.union1d(set_a,set_b)
print('Union of a and b:',union)
```

Array Attribute Functions

```python
a=np.array([1,2,3])
shape=a.shape # Shape of Array
size=a.size # Number of elements
dimensions=a.ndim # Number of dimensions
dtype=a.dtype # Data type of array

print('Shape of a:',shape)
print('Size of a:',size)
print('Dimensions of a:',dimensions)
print('Data Type of a:',dtype)
```

Other Functions

```python
# create a copy of an Array
a=np.array([1,2,3])
copied_array=np.copy(a)
print('Copied array:',copied_array)
```

```python
# Size of bytes in array
array_size_in_bytes=a.nbytes
print('Size of a in Bytes:',array_size_in_bytes)
```

```python
# Check if two arrays share memory
shared=np.shares_memory(a,copied_array)
print('Do a and copied_array share memory:',shared)
```
