Seaborn - Static Visualization (Advanced Plot)

```python
import matplotlib.pyplot as plt # visualization
import seaborn as sns # distribution visualization

%matplotlib inline
plt.rcParams['figure.figsize'] = 6,2

import warnings
warnings.filterwarnings('ignore') # os error
```

```python
df.head()
```

```python
df['InternetUsers']
```

```python
vis1 = sns.distplot(df['InternetUsers'])
plt.show()
```

```python
vis2 = sns.distplot(df['InternetUsers'])
plt.show()
```

```python
vis3 = sns.distplot(df['InternetUsers'], bins=10)
plt.show()
```

```python
vis4 = sns.boxplot(data = df, x="IncomeGroup", y="BirthRate") # Bi Variate Analysis
plt.show()
```

```python
vis5 = sns.lmplot(data = df, x="InternetUsers", y="BirthRate")
plt.show()
```

```python
vis6 = sns.lmplot(data = df, x="InternetUsers", y="BirthRate", fit_reg=False)
plt.show()
```

```python
vis6 = sns.lmplot(data = df, x="InternetUsers", y="BirthRate", fit_reg=True)
plt.show()
```

```python
vis6 = sns.lmplot(data = df, x="InternetUsers", y="BirthRate", fit_reg=False, hue='IncomeGroup') # Parameter for color
plt.show()
```

```python
vis6 = sns.lmplot(data = df, x="InternetUsers", y="BirthRate", fit_reg=True, hue='IncomeGroup') # Parameter for color
plt.show()
```
